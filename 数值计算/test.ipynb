```python
import numpy as np

A = [[1, 0.4, 0.4],
     [0.4, 1, 0.8],
     [0.4, 0.8, 1]]

print('转置矩阵:')
A_T = np.transpose(A)
print(A_T)

print('对角矩阵:')
D = np.diag(np.diag(A))
print(D)
    
print('下三角矩阵:')
L = np.tril(A) - D
print(L)

print('上三角矩阵:')
U = np.triu(A) - D
print(U)


print('高斯-赛德尔迭代矩阵:')
B_G = np.linalg.inv(L + D) @ (-U)
print(B_G)

print('特征值:')
eigvals = np.linalg.eigvals(B_G)
print(eigvals)

print('谱半径:')
spectral_radius = max(abs(eigvals))
print("谱半径:", spectral_radius)


print('SVD分解:')
U_svd, D_svd, V_svd = np.linalg.svd(B_G, compute_uv=False)
print('U:')
print(U_svd)
print('D:')
print(D_svd)
print('V:')
print(V_svd)
```
