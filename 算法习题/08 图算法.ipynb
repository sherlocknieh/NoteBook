#### 8.2

重写 DFS 算法，用栈结构替代递归。

答：

- 用栈保存 "待访问" 节点, 出栈表示已访问; 
- 每出栈一个节点时将其邻节点入栈。
- 为了避免重复访问，用一个 visited 集合标记已经入过栈的节点。
- 为了能处理非连通图和弱连通图，需要尝试用每个顶点作为起点遍历一次。


```python
# graph 是邻接表
def dfs_iterative(graph):

    result = []
    visited = set()

    # 尝试以每个顶点作为起点遍历一次
    for start in graph:
        # 跳过已访问过的节点
        if start in visited: continue

        stack = [start]     # 起点入栈
        visited.add(start)  # 标记起点已入栈

        while stack:
            u = stack.pop() # 取出栈顶
            result.append(u)# 访问 u
            
            neighbors = graph[u] # 获取 u 的邻居列表
            # 邻节点逆序入栈
            for v in reversed(neighbors):
                if v not in visited:
                    stack.append(v)
                    visited.add(v)
    return result

def test():
    # 测试用例
    graph = { 0: [1, 2], 1: [3], 2: [3], 3: [] }
    # 遍历图
    result = dfs_iterative(graph)
    print(result)  # 预期输出: [0, 1, 3, 2]

test()
```

#### 8.5

给出一个算法来确定一给定无向图 $G=(V,E)$ 中是否包含环，要求时间复杂度为 O(V)，与 E 无关。

答:

使用遍历法检测环的时间复杂度是 $O(V+E)$。

可以在遍历之前检查边和节点数关系, 若 $E \geq V$，则必定存在环。

当 $E < V$ 时, 遍历的时间复杂度变为 $O(2V)$ = $O(V)$。


```python
# graph: 邻接表
def has_cycle(graph):
    
    # 计算节点数和边数
    V = len(graph)
    E = sum(len(neighbors) for neighbors in graph.values()) // 2  # 无向图每条边被计数两次

    print(f"节点数: {V}, 边数: {E}")

    if E >= V:
        return True  # 边数大于等于节点数，必有环
    else:
        # 边数小于节点数，可能无环，需进一步检查
        visited = set()
        def dfs(u, parent):
            visited.add(u)
            for v in graph[u]:
                if v not in visited:
                    if dfs(v, u):
                        return True
                elif v != parent:
                    return True
            return False

        # 对每个顶点尝试作为起点, 进行 DFS 检测环
        for vertex in graph:
            if vertex not in visited:
                if dfs(vertex, None):
                    return True

        return False

def test():
    # 有环测试用例
    graph = {
        'A': ['B', 'C'],
        'B': ['A', 'D'],
        'C': ['A', 'D'],
        'D': ['B', 'C']
    }
    assert has_cycle(graph) == True
    print("测试通过!")

    # 无环测试用例
    graph = {
        'A': ['B', 'C'],
        'B': ['A'],
        'C': ['A']
    }
    assert has_cycle(graph) == False
    print("测试通过!")

test()
```
