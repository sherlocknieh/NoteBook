## 最大流问题


初始管道容量:
```mermaid
graph LR
    S(("S"))
    A(("A"))
    B(("B"))
    T(("T"))

    S ==1==> A
    A ==1==> T
    A ==1==> B
    S ==1==> B

    B ==1==> T
```


````python
import networkx as nx
import matplotlib.pyplot as plt

# 1. 创建有向图
G = nx.DiGraph()

# 2. 添加边（自动创建节点）
G.add_edges_from([("A", "B"), ("B", "C"), ("C", "A"), ("A", "D")])

# 3. 绘制并输出图像
nx.draw(G, with_labels=True, node_color='lightblue', font_weight='bold')
plt.savefig("graph.png") # 输出为图片
plt.show()
````

````python
from pathlib import Path
from html import escape

from IPython.display import HTML, display
from pyvis.network import Network



import matplotlib.pyplot as plt
import networkx as nx
from IPython.display import Image, display
nt = Network(height='550px', width='100%', directed=True, notebook=True, cdn_resources='in_line')
nt.from_nx(G)
html_content = nt.generate_html()
filename = 'karate.html'
Path(filename).write_text(html_content, encoding='utf-8')
# notebook 里优先显示稳定的静态图，交互版保留在 karate.html
pos = nx.spring_layout(G, seed=42)
plt.figure(figsize=(7, 5))
nx.draw(G, pos, with_labels=True, node_color='lightblue', font_weight='bold', arrows=True)
plt.tight_layout()
plt.savefig('karate.png', dpi=160, bbox_inches='tight')
plt.close()
display(Image(filename='karate.png'))
````
